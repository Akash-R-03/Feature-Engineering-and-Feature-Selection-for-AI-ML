# 1. What is a Feature?

A feature is an individual measurable property or characteristic of a dataset that is used as an input to a Machine Learning model.

Examples from the Titanic dataset:

- Pclass
- Sex
- Age
- SibSp
- Parch
- Fare
- Embarked

For example, Age represents the age of a passenger.

### AI/ML Relevance

Features provide the information from which a Machine Learning model learns patterns and makes predictions.

```python
feature_columns = [
    "Pclass",
    "Sex",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "Embarked"
]

print("Example Features:")
print(df[feature_columns].head())
```

# 2. Feature vs Variable

A variable is a general term for an attribute or value recorded in a dataset.

A feature is a variable that is selected or prepared as an input for a Machine Learning model.

Example:

- Age is a variable in the dataset.
- When Age is used as an input to predict Survived, it becomes a feature.

Therefore, a variable does not automatically have to be used as a feature.

```python
print(df[["Age", "Fare", "Pclass"]].head())
```

# 3. Feature vs Target

A feature is an input given to the Machine Learning model.

A target is the output that the model is trying to predict.

For the Titanic dataset:

Features:
- Pclass
- Sex
- Age
- SibSp
- Parch
- Fare
- Embarked

Target:
- Survived

### AI/ML Relevance

Correctly separating features and the target is essential for supervised Machine Learning.

```python
features = df[
    ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]
]

target = df["Survived"]

print("Feature Shape:", features.shape)
print("Target Shape:", target.shape)
```

# 4. What is Feature Engineering?

Feature Engineering is the process of creating, transforming, or extracting meaningful information from existing data so that Machine Learning models can learn useful patterns more effectively.

It is not simply about creating more columns.

A good engineered feature should:

- Have a meaningful purpose
- Represent useful information
- Help the model learn relevant patterns
- Be based on valid data
- Avoid feature leakage

### Titanic Example

SibSp and Parch contain family-related information.

We can combine them with the passenger themselves to create:

FamilySize = SibSp + Parch + 1

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

print(df[["SibSp", "Parch", "FamilySize"]].head())
```

# 5. Why Feature Engineering?

Raw data does not always represent information in the most useful form for Machine Learning.

Feature Engineering can:

- Combine related information
- Create meaningful representations
- Capture useful relationships
- Improve model learning
- Make features easier to interpret

### Example

SibSp and Parch separately describe family relationships.

FamilySize combines them into one feature representing the total travelling family size.

### AI/ML Relevance

A meaningful feature can make an important pattern easier for a Machine Learning model to learn.

```python
print(df[["SibSp", "Parch", "FamilySize"]].head(10))
```

# 6. Importance of Domain Knowledge

Domain knowledge means understanding the real-world meaning of the data.

Feature Engineering should not be based only on mathematical operations.

For the Titanic dataset:

- SibSp represents siblings/spouses aboard.
- Parch represents parents/children aboard.
- FamilySize represents the total family group travelling with the passenger.

Understanding these meanings helps us create features that have a logical purpose.

### AI/ML Relevance

Domain knowledge helps an AI/ML Engineer create features that are meaningful, useful, and relevant to the problem.

# 7. Good Features vs Bad Features

A good feature:

- Has a clear meaning
- Uses relevant source information
- Has potential ML value
- Is available at prediction time
- Does not introduce leakage

A bad feature may be:

- Irrelevant
- Redundant
- Random
- Extremely noisy
- Based on information unavailable at prediction time
- A source of data leakage

### Example

FamilySize is a meaningful feature because it is calculated from relevant passenger information.

A randomly generated number would not have meaningful business or ML value.

# 9. Feature Creation

Feature Creation means creating a new feature from one or more existing columns.

### Example

FamilySize = SibSp + Parch + 1

### Required Feature Documentation

**Feature Name:** FamilySize

**Source Columns:** SibSp, Parch

**Logic:** SibSp + Parch + 1

**Reason:** To represent the total number of people in the passenger's travelling family group.

**Business Meaning:** Represents the size of the passenger's family group during the journey.

**ML Relevance:** Family size may provide useful information about passenger survival patterns.

**Leakage Check:** No. SibSp and Parch are available passenger information and do not use the target.

**Final Decision:** Retain

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

print(df[["SibSp", "Parch", "FamilySize"]].head())
```

# 10. Feature Creation – IsAlone

We can create another feature from FamilySize.

### Required Feature Documentation

**Feature Name:** IsAlone

**Source Columns:** FamilySize

**Logic:** IsAlone = 1 when FamilySize equals 1, otherwise 0.

**Reason:** To distinguish passengers travelling alone from passengers travelling with family.

**Business Meaning:** Represents whether the passenger was travelling alone.

**ML Relevance:** Travelling alone may provide useful information for predicting survival.

**Leakage Check:** No. The feature uses only passenger information and does not use Survived.

**Final Decision:** Retain

```python
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print(df[["FamilySize", "IsAlone"]].head(10))
```

# 11. Feature Creation – FarePerPerson

We can create a feature representing the fare amount relative to the travelling family size.

### Formula

FarePerPerson = Fare / FamilySize

### Required Feature Documentation

**Feature Name:** FarePerPerson

**Source Columns:** Fare, FamilySize

**Logic:** Fare divided by FamilySize.

**Reason:** To represent the approximate fare contribution per person in the travelling family group.

**Business Meaning:** Provides a normalized view of fare relative to family size.

**ML Relevance:** It may provide information that is different from using total Fare alone.

**Leakage Check:** No. It does not use the target variable.

**Final Decision:** Needs Further Analysis

```python
df["FarePerPerson"] = df["Fare"] / df["FamilySize"]

print(df[["Fare", "FamilySize", "FarePerPerson"]].head())
```

# 12. Feature Transformation

Feature Transformation changes the representation of an existing feature.

Common transformations include:

- Log transformation
- Square-root transformation
- Standardization
- Normalization
- Binning

### Titanic Example

Fare can contain large values and may be skewed.

A log transformation can reduce the influence of very large values.

### Required Feature Documentation

**Feature Name:** Fare_Log

**Source Columns:** Fare

**Logic:** log1p(Fare)

**Reason:** To reduce the effect of extreme values and provide a transformed representation of Fare.

**Business Meaning:** Represents Fare on a compressed logarithmic scale.

**ML Relevance:** May help algorithms that are sensitive to feature distributions.

**Leakage Check:** No. Only Fare is used.

**Final Decision:** Needs Further Analysis

```python
df["Fare_Log"] = np.log1p(df["Fare"])

print(df[["Fare", "Fare_Log"]].head())
```

# 13. Feature Extraction

Feature Extraction means obtaining useful information from an existing complex or raw feature.

### Titanic Example

The Name column contains passenger titles such as Mr, Mrs, Miss, etc.

Instead of using the complete name directly, we can extract the title as a new feature.

### Required Feature Documentation

**Feature Name:** Title

**Source Columns:** Name

**Logic:** Extract the text between the comma and period in the passenger name.

**Reason:** To convert useful information contained in the Name column into a structured feature.

**Business Meaning:** Represents the passenger's title.

**ML Relevance:** Titles can represent differences in social status, age group, or gender-related information.

**Leakage Check:** No. The title is extracted from the passenger's name and does not use Survived.

**Final Decision:** Retain

```python
df["Title"] = df["Name"].str.extract(r",\s*([^.]*)\.", expand=False)

print(df[["Name", "Title"]].head(10))
```

# 14. Feature Selection

Feature Selection is the process of choosing useful features and removing features that are unnecessary, redundant, irrelevant, or risky.

Feature Engineering and Feature Selection are different:

- Feature Engineering creates or transforms features.
- Feature Selection chooses which features should be retained.

### Example

After creating FamilySize, IsAlone, FarePerPerson, Fare_Log, and Title, we should evaluate whether each feature provides useful information.

A feature should not be retained only because it is newly created.

### AI/ML Relevance

Feature Selection can reduce unnecessary model complexity, improve interpretability, and reduce the effect of irrelevant information.

# 15. Feature Leakage

Feature Leakage occurs when a feature contains information that would not legitimately be available when the prediction is made.

Leakage can produce unrealistically high model performance.

### Example

If a feature is created using the target variable `Survived`, it would be leakage.

For example:

SurvivalRate = group-based survival information calculated using the target

This should not be used as a normal input feature without a leakage-safe process.

### Prevention

- Do not use target information when creating predictive features.
- Understand when the source information becomes available.
- Perform feature engineering without using future information.
- Apply transformations using training data where appropriate.

### AI/ML Relevance

Preventing leakage ensures that model evaluation represents real-world performance.

```python
target = "Survived"

engineered_features = [
    "FamilySize",
    "IsAlone",
    "FarePerPerson",
    "Fare_Log",
    "Title"
]

print("Target:", target)
print("Engineered Features:", engineered_features)

print("\nTarget used directly in engineered features:",
      target in engineered_features)
```
