## 1. Text Length

Text length measures the total number of characters in a text value.

### Feature Name
`Name_Text_Length`

### Source Column
- `Name`

### Logic
Count the total number of characters in the passenger name.

### Reason
Text length can provide information about the structure and complexity of text.

### Business Meaning
In customer reviews, longer text may indicate more detailed feedback.

### ML Relevance
Text length can help a model distinguish between short and detailed text.

### Leakage Check
The target variable is not used.

### Final Decision
**Retain**

```python
df["Name_Text_Length"] = df["Name"].fillna("").str.len()

df[["Name", "Name_Text_Length"]].head(10)
```

Count sentence-ending punctuation such as `.`, `!`, and `?`.

### Business Meaning

No target information is used.

**Needs Further Analysis**

## 10. Bag of Words

Bag of Words (BoW) converts text into numerical features based on word occurrence.

It creates a vocabulary from the available documents.

Each document is represented by the number of times each vocabulary word appears.

### Example

Documents:

1. `good product`
2. `good service`
3. `bad service`

Possible vocabulary:

- good
- product
- service
- bad

Each document can then be represented as a numerical vector.

### Advantages

- Simple
- Easy to understand
- Useful for basic text classification

### Limitations

- Ignores word order
- Can create many features
- Produces sparse matrices
- Does not capture semantic meaning

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "good product",
    "good service",
    "bad service",
    "good product service"
]

vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(documents)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

bow_df
```

## 11. TF-IDF

TF-IDF stands for Term Frequency–Inverse Document Frequency.

It assigns higher importance to words that are frequent in a particular document but less common across all documents.

### Main Idea

A word that appears in every document is usually less informative.

A word that appears in only a few documents may be more useful for distinguishing those documents.

### Advantages

- Reduces importance of very common words
- Useful for text classification
- Produces numerical text representations

### Limitations

- Does not understand semantic meaning
- Can produce many features
- Requires suitable preprocessing

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

tfidf_df.round(3)
```

## 12. N-Grams

N-Grams are sequences of consecutive words or tokens.

Types include:

- Unigram → one word
- Bigram → two words
- Trigram → three words

### Example

Text:

`very good product`

Unigrams:

- very
- good
- product

Bigrams:

- very good
- good product

Trigrams:

- very good product

### Why Use N-Grams?

N-Grams preserve some word-order information that Bag of Words loses.

For example:

`not good`

has a different meaning from:

`good`

A bigram can capture the phrase `not good`.

### Limitation

Higher-order N-Grams can significantly increase the number of features and model complexity.

```python
bigram_vectorizer = CountVectorizer(ngram_range=(1, 2))

bigram_matrix = bigram_vectorizer.fit_transform(documents)

bigram_df = pd.DataFrame(
    bigram_matrix.toarray(),
    columns=bigram_vectorizer.get_feature_names_out()
)

bigram_df
```

## Advantages and Limitations of Text Feature Engineering

### Advantages

- Converts raw text into ML-readable numerical features
- Captures useful information from text
- Supports text classification
- Can identify important keywords
- Can represent word occurrence and phrase patterns

### Limitations

- Text can create a very large number of features
- Bag of Words ignores word order
- TF-IDF does not understand semantic meaning
- N-Grams can increase dimensionality
- Text preprocessing may be required
- Poorly selected features can create noise
- Data leakage can occur if the text vectorizer is fitted using the complete dataset

### Final Conclusion

Raw text cannot usually be given directly to traditional Machine Learning models.

Text feature engineering converts raw text into numerical representations such as counts, Bag of Words, TF-IDF, and N-Grams.

The appropriate technique depends on the dataset, business problem, amount of text, model, and computational requirements.
