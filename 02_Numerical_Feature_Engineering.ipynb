```python
import pandas as pd
import numpy as np

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset Shape:", df.shape)
print(df[["Age", "SibSp", "Parch", "Fare", "Pclass"]].head())
```

# 1. Mathematical Transformations

Mathematical transformations create new numerical features by applying mathematical operations to existing numerical columns.

Common operations include:

- Addition
- Subtraction
- Multiplication
- Division

### When to Use

Use mathematical transformations when two or more numerical variables have a meaningful mathematical relationship.

### Why Use Them?

They can combine information and create a feature that represents a more useful relationship than the original columns alone.

### AI/ML Relevance

A meaningful mathematical feature can make relationships in the data easier for a Machine Learning model to learn.

# 2. Addition

Addition combines numerical values.

### Titanic Example

FamilySize = SibSp + Parch + 1

The `1` represents the passenger themselves.

### Required Documentation

**Feature Name:** FamilySize

**Source Columns:** SibSp, Parch

**Logic:** SibSp + Parch + 1

**Reason:** To combine family-related information into one feature.

**Business Meaning:** Represents the total number of people in the passenger's travelling family group.

**ML Relevance:** Family size may provide useful information for predicting survival.

**Leakage Check:** No. It does not use the target variable.

**Final Decision:** Retain

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

print(df[["SibSp", "Parch", "FamilySize"]].head())
```

# 3. Subtraction

Subtraction calculates the difference between two numerical values.

### Titanic Example

We can calculate:

SiblingParentDifference = SibSp - Parch

### When to Use

Use subtraction when the difference between two measurements has a meaningful interpretation.

### Required Documentation

**Feature Name:** SiblingParentDifference

**Source Columns:** SibSp, Parch

**Logic:** SibSp - Parch

**Reason:** To represent the difference between sibling/spouse count and parent/child count.

**Business Meaning:** Shows whether a passenger had relatively more siblings/spouses or parents/children travelling with them.

**ML Relevance:** May provide an additional representation of family structure.

**Leakage Check:** No.

**Final Decision:** Needs Further Analysis

```python
df["SiblingParentDifference"] = df["SibSp"] - df["Parch"]

print(df[["SibSp", "Parch", "SiblingParentDifference"]].head())
```

# 4. Multiplication

Multiplication combines numerical values by calculating their product.

### Titanic Example

We can create:

AgeClassInteraction = Age × Pclass

### When to Use

Use multiplication when the interaction between two numerical variables may carry useful information.

### Required Documentation

**Feature Name:** AgeClassInteraction

**Source Columns:** Age, Pclass

**Logic:** Age × Pclass

**Reason:** To represent the combined effect of passenger age and passenger class.

**Business Meaning:** Represents an interaction between age and travel class.

**ML Relevance:** May help a model capture an interaction between two variables.

**Leakage Check:** No.

**Final Decision:** Needs Further Analysis

```python
df["AgeClassInteraction"] = df["Age"] * df["Pclass"]

print(df[["Age", "Pclass", "AgeClassInteraction"]].head())
```

# 5. Division

Division creates a feature by dividing one numerical value by another.

### Example

FarePerPerson = Fare / FamilySize

### When to Use

Use division when the absolute value alone is less informative than the value relative to another measurement.

### Required Documentation

**Feature Name:** FarePerPerson

**Source Columns:** Fare, FamilySize

**Logic:** Fare / FamilySize

**Reason:** To represent the approximate fare per person.

**Business Meaning:** Shows the fare amount relative to the passenger's family group size.

**ML Relevance:** Provides a normalized representation of fare that may capture information different from total Fare.

**Leakage Check:** No.

**Final Decision:** Needs Further Analysis

```python
df["FarePerPerson"] = df["Fare"] / df["FamilySize"]

print(df[["Fare", "FamilySize", "FarePerPerson"]].head())
```

# 6. Ratios

### Titanic Example

ParchRatio = Parch / FamilySize

### When to Use

Use ratios when relative proportions are more useful than absolute values.

### Required Documentation

**Feature Name:** ParchRatio

**Source Columns:** Parch, FamilySize

**Logic:** Parch / FamilySize

**Reason:** To represent the proportion of the travelling family group accounted for by parents/children.

**Business Meaning:** Shows the relative contribution of Parch to family size.

**ML Relevance:** Provides proportional family information instead of only raw counts.

**Leakage Check:** No.

**Final Decision:** Needs Further Analysis

```python
df["ParchRatio"] = df["Parch"] / df["FamilySize"]

print(df[["Parch", "FamilySize", "ParchRatio"]].head())
```

# 7. Percentages

A percentage expresses a value relative to a total on a 0–100 scale.

### Titanic Example

ParchPercentage = (Parch / FamilySize) × 100

### When to Use

Use percentages when the relative contribution of a value is easier to understand than its raw count.

### Required Documentation

**Feature Name:** ParchPercentage

**Source Columns:** Parch, FamilySize

**Logic:** (Parch / FamilySize) × 100

# 8. Differences

We can compare the number of siblings/spouses and parents/children.

# 9. Aggregations

Aggregation combines or summarizes numerical information.

Examples include:

- Sum
- Mean
- Minimum
- Maximum
- Count

### Titanic Example

FamilySize is an aggregation of family-related counts.

Another useful aggregation is average fare per family member.

### When to Use

Use aggregation when multiple numerical values represent different parts of the same concept.

### Why Use It?

Aggregation reduces multiple related values into a meaningful summary.

# 10. Log Transformation

Log transformation changes a numerical feature using a logarithmic function.

Formula:

log(1 + x)

`log1p()` is used so that zero values can be handled safely.

### When to Use

Log transformation is useful when:

- Data is highly right-skewed
- A feature has very large values
- We want to reduce the influence of extreme values

### Titanic Example

Fare has a wide range and can be skewed.

### Required Documentation

**Feature Name:** Fare_Log

**Source Columns:** Fare

**Logic:** log1p(Fare)

**Reason:** To reduce the effect of very large fare values.

**Business Meaning:** Represents Fare on a compressed logarithmic scale.

**ML Relevance:** May help models handle skewed numerical distributions.

**Leakage Check:** No.

**Final Decision:** Needs Further Analysis

```python
df["Fare_Log"] = np.log1p(df["Fare"])

print(df[["Fare", "Fare_Log"]].head())
```

# 11. Square Root Transformation

Square root transformation applies the square root to a numerical feature.

Formula:

√x

### When to Use

It can be useful for reducing moderate positive skewness and compressing large values.

### Titanic Example

We can apply it to Fare.

### Required Documentation

**Feature Name:** Fare_Sqrt

**Source Columns:** Fare

**Logic:** √Fare

**Reason:** To reduce the influence of large Fare values.

**Business Meaning:** Represents Fare on a square-root-transformed scale.

**ML Relevance:** May improve the representation of a skewed numerical feature.

**Leakage Check:** No.

**Final Decision:** Needs Further Analysis

```python
df["Fare_Sqrt"] = np.sqrt(df["Fare"])

print(df[["Fare", "Fare_Sqrt"]].head())
```

### When to Use

### Required Documentation

**Feature Name:** Age_Squared

# 13. Absolute Difference

Absolute difference measures the size of the difference without considering its direction.

Formula:

|SibSp - Parch|

### When to Use

Use absolute difference when only the magnitude of the difference matters.

### Required Documentation

**Feature Name:** AbsoluteFamilyDifference

**Source Columns:** SibSp, Parch

**Logic:** |SibSp - Parch|

**Reason:** To measure how different the two family relationship counts are without considering which one is larger.

**Business Meaning:** Represents the magnitude of the difference between siblings/spouses and parents/children.

**ML Relevance:** Provides a simple numerical measure of family structure difference.

**Leakage Check:** No.

**Final Decision:** Needs Further Analysis

```python
df["AbsoluteFamilyDifference"] = abs(df["SibSp"] - df["Parch"])

print(df[["SibSp", "Parch", "AbsoluteFamilyDifference"]].head())
```

# 14. Relative Difference

Relative difference represents a difference relative to a reference value.

A safe formula is:

Relative Difference = |A - B| / (A + B)

A small constant is added when required to avoid division by zero.

### Titanic Example

We compare SibSp and Parch.

### Required Documentation

**Feature Name:** RelativeFamilyDifference

**Source Columns:** SibSp, Parch

**Logic:** |SibSp - Parch| / (SibSp + Parch)

**Reason:** To measure the difference relative to the total family relationship count.

**Business Meaning:** Represents the relative imbalance between the two family relationship counts.

**ML Relevance:** Provides proportional information about family structure.

**Leakage Check:** No.

**Final Decision:** Needs Further Analysis

```python
denominator = df["SibSp"] + df["Parch"]

df["RelativeFamilyDifference"] = (
    abs(df["SibSp"] - df["Parch"]) /
    denominator.replace(0, np.nan)
)

print(df[["SibSp", "Parch", "RelativeFamilyDifference"]].head())
```

# 16. Normalized Features

Normalization changes the scale of a numerical feature to a common range.

A common Min-Max formula is:

Normalized Value = (x - min(x)) / (max(x) - min(x))

This normally produces values between 0 and 1.

### When to Use

Normalization can be useful when:

- Features have very different numerical ranges
- The Machine Learning algorithm is sensitive to feature scale
- We want values represented on a common scale

### Important ML Practice

When preparing data for a real Machine Learning model, the minimum and maximum values should be learned from the training data only.

### Required Documentation

**Feature Name:** Fare_Normalized

**Source Columns:** Fare

**Logic:** (Fare - minimum Fare) / (maximum Fare - minimum Fare)

**Reason:** To represent Fare on a common 0–1 scale.

**Business Meaning:** Represents the relative position of a passenger's Fare within the dataset range.

**ML Relevance:** Can make Fare comparable with other numerical features for scale-sensitive algorithms.

**Leakage Check:** If calculated using the complete dataset before splitting, it can cause preprocessing leakage. In a real ML workflow, fit normalization only on training data.

**Final Decision:** Needs Further Analysis

```python
fare_min = df["Fare"].min()
fare_max = df["Fare"].max()

df["Fare_Normalized"] = (
    (df["Fare"] - fare_min) /
    (fare_max - fare_min)
)

print(df[["Fare", "Fare_Normalized"]].head())
```
