# Notebook 4 – Date & Time Feature Engineering

## Objective

Date and Time Feature Engineering converts date and time information into useful numerical or categorical features that Machine Learning models can understand.

Topics covered:

- Year
- Month
- Day
- Day of Week
- Week Number
- Quarter
- Weekend Indicator
- Month Start
- Month End
- Days Since Event
- Time Difference
- Customer Age
- Customer Tenure
- Recency
- Frequency
- Seasonality

Examples:

- Customer age
- Employee tenure
- Days since last purchase
- Days between transactions
- Purchase month
- Weekend purchase indicator

Dataset: Titanic-Dataset.csv

Note:
The Titanic dataset does not contain customer transaction dates or employee dates. Therefore, a synthetic event date is created only to demonstrate Date & Time Feature Engineering.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("Titanic-Dataset.csv")

# Create a synthetic event date for demonstration
base_date = pd.Timestamp("1912-04-01")

df["EventDate"] = (
    base_date
    + pd.to_timedelta(df.index % 30, unit="D")
)

print("Dataset Shape:", df.shape)
print(df[["PassengerId", "EventDate"]].head())
```

# 1. Year

Year extracts the year from a date.

### Example

EventDate → 1912

### When to Use

Use Year when the year itself may contain useful information, such as:

- Business trends
- Historical patterns
- Policy periods
- Product lifecycle

### ML Relevance

Year can help a model identify long-term trends or differences between time periods.

```python
df["Event_Year"] = df["EventDate"].dt.year

print(df[["EventDate", "Event_Year"]].head())
```

# 2. Month

Month extracts the month number from a date.

### Example

April → 4

### When to Use

Use Month when the behavior or target may change during different months.

### ML Relevance

Month can capture monthly patterns and seasonality.

```python
df["Event_Month"] = df["EventDate"].dt.month

print(df[["EventDate", "Event_Month"]].head())
```

# 3. Day

Day extracts the day of the month from a date.

### When to Use

Day can be useful when behavior changes based on the position of the day within a month.

### Caution

Day alone may not always provide useful information because day 1, day 2, etc. can repeat every month.

```python
df["Event_Day"] = df["EventDate"].dt.day

print(df[["EventDate", "Event_Day"]].head())
```

# 4. Day of Week

Day of Week represents the day from Monday to Sunday.

Monday = 0
Tuesday = 1
...
Sunday = 6

### When to Use

Useful when behavior differs between weekdays and weekends.

### ML Relevance

It can capture weekly behavioral patterns.

```python
df["Day_Of_Week"] = df["EventDate"].dt.dayofweek

print(df[["EventDate", "Day_Of_Week"]].head())
```

# 5. Week Number

Week Number identifies the week of the year.

### When to Use

Useful for:

- Weekly sales analysis
- Weekly customer activity
- Operational trends
- Time-series modeling

### ML Relevance

It can help capture weekly patterns.

```python
df["Week_Number"] = df["EventDate"].dt.isocalendar().week.astype(int)

print(df[["EventDate", "Week_Number"]].head())
```

# 6. Quarter

Quarter divides a year into four periods:

Q1 → January to March
Q2 → April to June
Q3 → July to September
Q4 → October to December

### When to Use

Useful for quarterly business and financial analysis.

### ML Relevance

Quarter can capture seasonal or business-cycle patterns.

```python
df["Quarter"] = df["EventDate"].dt.quarter

print(df[["EventDate", "Quarter"]].head())
```

# 7. Weekend Indicator

Weekend Indicator identifies whether an event occurred on Saturday or Sunday.

Logic:

Weekend = 1
Weekday = 0

### When to Use

Useful when customer behavior differs between weekdays and weekends.

### ML Relevance

It can help a model learn weekday-versus-weekend behavior.

```python
df["Weekend_Indicator"] = (
    df["EventDate"].dt.dayofweek >= 5
).astype(int)

print(df[["EventDate", "Weekend_Indicator"]].head(10))
```

# 8. Month Start

Month Start identifies whether a date is the first day of its month.

### When to Use

Useful when business activity changes around the beginning of a month.

Examples:

- Salary processing
- Monthly subscriptions
- Monthly reporting

```python
df["Month_Start"] = df["EventDate"].dt.is_month_start.astype(int)

print(df[["EventDate", "Month_Start"]].head())
```

# 9. Month End

Month End identifies whether a date is the last day of its month.

### When to Use

Useful for:

- Monthly closing
- Billing
- Financial reporting
- Subscription cycles

```python
df["Month_End"] = df["EventDate"].dt.is_month_end.astype(int)

print(df[["EventDate", "Month_End"]].head())
```

# 10. Days Since Event

Days Since Event measures how many days have passed since a particular event.

### Formula

Days Since Event = Reference Date - Event Date

### When to Use

Useful for:

- Customer activity
- Last purchase
- Last login
- Last interaction

### ML Relevance

Recent activity can be an important predictor of future customer behavior.

```python
reference_date = pd.Timestamp("1912-05-15")

df["Days_Since_Event"] = (
    reference_date - df["EventDate"]
).dt.days

print(df[["EventDate", "Days_Since_Event"]].head())
```

# 11. Required Documentation – Days Since Event

**Feature Name:** Days_Since_Event

**Source Columns:** EventDate

**Logic:** Reference Date - EventDate.

**Reason:** Measure how recently an event occurred.

**Business Meaning:** Represents the number of days since the event.

**ML Relevance:** Recency can help models understand recent versus older activity.

**Leakage Check:** The reference date must represent a valid prediction-time date. Future information must not be used.

**Final Decision:** Retain

# 12. Time Difference

Time Difference measures the amount of time between two dates.

### Formula

Time Difference = End Date - Start Date

### Example

Days between order date and delivery date.

### When to Use

Useful for:

- Delivery duration
- Processing time
- Customer response time
- Employee service duration

```python
df["EndDate"] = df["EventDate"] + pd.to_timedelta(
    (df.index % 7) + 1,
    unit="D"
)

df["Time_Difference_Days"] = (
    df["EndDate"] - df["EventDate"]
).dt.days

print(
    df[
        ["EventDate", "EndDate", "Time_Difference_Days"]
    ].head()
)
```

# 13. Customer Age

Customer Age represents the age of a customer at a particular event or reference date.

### Formula

Customer Age = Event Date - Date of Birth

### When to Use

Useful when age may influence:

- Customer behavior
- Product preference
- Risk
- Purchasing decisions

### Titanic Example

The Titanic dataset already contains `Age`, so we can use it as an example of an age-related feature.

### Important Note

This is not calculated from a real Date of Birth column because the Titanic dataset does not contain passenger birth dates.

```python
df["Customer_Age"] = df["Age"]

print(df[["Age", "Customer_Age"]].head())
```

# 14. Required Documentation – Customer Age

**Feature Name:** Customer_Age

**Source Columns:** Age

**Logic:** Use the available age value as the age at the event.

**Reason:** Represent age as a directly usable numerical feature.

**Business Meaning:** Represents the passenger's age.

**ML Relevance:** Age can help a model identify age-related patterns in the target.

**Leakage Check:** No target information is used.

**Final Decision:** Retain

# 15. Customer Tenure

Customer Tenure represents the amount of time a customer has been associated with a company or service.

### Formula

Customer Tenure = Reference Date - Customer Start Date

### Example

If a customer joined on January 1 and the reference date is January 31:

Tenure = 30 days

### When to Use

Useful for:

- Customer churn prediction
- Customer segmentation
- Retention analysis
- Subscription analysis

### Titanic Dataset

The Titanic dataset does not contain customer joining dates. Therefore, this concept is demonstrated using synthetic dates only.

```python
df["Customer_Start_Date"] = df["EventDate"] - pd.to_timedelta(
    (df.index % 100) + 10,
    unit="D"
)

df["Customer_Tenure_Days"] = (
    df["EventDate"] - df["Customer_Start_Date"]
).dt.days

print(
    df[
        [
            "Customer_Start_Date",
            "EventDate",
            "Customer_Tenure_Days"
        ]
    ].head()
)
```

# 16. Required Documentation – Customer Tenure

**Feature Name:** Customer_Tenure_Days

**Source Columns:** Customer_Start_Date, EventDate

**Logic:** EventDate - Customer_Start_Date.

**Reason:** Measure the duration of the customer's relationship with the service.

**Business Meaning:** Represents customer tenure in days.

**ML Relevance:** Tenure can be useful for predicting churn, retention, and customer behavior.

**Leakage Check:** Start date and reference date must be available at prediction time.

**Final Decision:** Needs Further Analysis

# 17. Recency

Recency measures how recently a customer's previous activity occurred.

### Formula

Recency = Current Reference Date - Last Activity Date

Lower recency means the activity happened more recently.

### When to Use

Recency is commonly used in:

- Customer analytics
- Churn prediction
- Marketing
- Recommendation systems

### ML Relevance

Recent activity may be strongly related to future customer behavior.

```python
df["Last_Activity_Date"] = df["EventDate"] - pd.to_timedelta(
    (df.index % 15),
    unit="D"
)

df["Recency_Days"] = (
    df["EventDate"] - df["Last_Activity_Date"]
).dt.days

print(
    df[
        ["EventDate", "Last_Activity_Date", "Recency_Days"]
    ].head()
)
```

# 18. Required Documentation – Recency

**Feature Name:** Recency_Days

**Source Columns:** EventDate, Last_Activity_Date

**Logic:** EventDate - Last_Activity_Date.

**Reason:** Measure how recently the previous activity occurred.

**Business Meaning:** Represents the number of days since the customer's previous activity.

**ML Relevance:** Recency can be useful for predicting future activity or customer behavior.

**Leakage Check:** Last activity must occur before the prediction point. Future activity must not be used.

**Final Decision:** Retain

# 19. Frequency

Frequency measures how often an event occurs within a specific time period.

### Example

Number of purchases made by a customer during the previous 30 days.

### When to Use

Useful for:

- Customer segmentation
- Churn prediction
- Recommendation systems
- Purchase behavior analysis

### ML Relevance

Customers with different activity frequencies may have different future behavior.

```python
# Demonstration using Ticket as a grouping identifier
df["Transaction_Frequency"] = df.groupby("Ticket")["Ticket"].transform("count")

print(
    df[
        ["Ticket", "Transaction_Frequency"]
    ].head(10)
)
```

# 20. Required Documentation – Frequency

**Feature Name:** Transaction_Frequency

**Source Columns:** Ticket

**Logic:** Count the number of records associated with each Ticket.

**Reason:** Demonstrate how event frequency can be calculated for a repeated group identifier.

**Business Meaning:** Represents how many passengers are associated with the same ticket value in this dataset.

**ML Relevance:** Frequency-based information can sometimes capture group size or activity patterns.

**Leakage Check:** In a real predictive problem, the frequency must be calculated only from information available before the prediction point. Future records must not be included.

**Final Decision:** Needs Further Analysis

# 21. Seasonality

Seasonality means repeated patterns that occur at regular time intervals.

Examples:

- Higher sales during festivals
- Increased travel during holidays
- Weekend purchasing patterns
- Seasonal demand

### Common Seasonal Features

- Month
- Quarter
- Day of Week
- Weekend Indicator

### ML Relevance

Seasonal features help models capture repeated time-based patterns.

```python
df["Season"] = df["Event_Month"].map({
    12: "Winter",
    1: "Winter",
    2: "Winter",
    3: "Spring",
    4: "Spring",
    5: "Spring",
    6: "Summer",
    7: "Summer",
    8: "Summer",
    9: "Autumn",
    10: "Autumn",
    11: "Autumn"
})

print(df[["EventDate", "Event_Month", "Season"]].head())
```

# 22. Required Documentation – Seasonality

**Feature Name:** Season

**Source Columns:** Event_Month

**Logic:** Map months into seasonal groups.

**Reason:** Capture recurring seasonal patterns.

**Business Meaning:** Represents the broad seasonal period associated with the event date.

**ML Relevance:** Seasonal information can help a model learn recurring time-based patterns.

**Leakage Check:** No leakage if the event date is known at prediction time.

**Final Decision:** Needs Further Analysis

# 25. Conclusion

Date & Time Feature Engineering converts raw date information into useful features that can represent:

- Time periods
- Trends
- Durations
- Recency
- Frequency
- Seasonality
- Customer age
- Customer tenure

The important principle is that a date itself may not be directly useful to a Machine Learning model. Extracting meaningful information such as month, weekday, quarter, recency, or duration can provide a better representation.

However, date-based features must always be checked for leakage.

For example, when predicting customer churn today, information from a customer's future activity must not be used to create the features.

Therefore, every date/time feature should be evaluated based on:

- Business meaning
- Availability at prediction time
- ML usefulness
- Time relationship
- Leakage risk

The goal is to transform dates into meaningful, reliable, and ML-ready features.
